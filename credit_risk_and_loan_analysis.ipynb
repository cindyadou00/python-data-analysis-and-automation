```python
# PASSWORD VALIDATION ANALYSIS

# 1. Password Validation with Detailed Error Messages
```

```python
# Function to check the validity of a password
def check_password(password):
    errors = []
    
    # the lenth of the password
    if len(password) < 12:
        errors.append("Password must be at least 12 characters long.")
    
    # uppercase
    if not any(char.isupper() for char in password):
        errors.append("Password must contain at least one uppercase letter.")
    
    # symbols
    if not any(char in "@()&" for char in password):
        errors.append("Password must contain at least one of the symbols: @, (, ), &.")
    
    # the errors
    if not errors:
        return "Valid"
    else:
        return "Invalid: " + ", ".join(errors)
```

```python
#my path
file_path = "data/Candidate_Passwords.csv"
```

```python
try:
    # Load CSV-Pandas DataFrame
    df = pd.read_csv(file_path)
    
    # Display the available columns
    print("Available columns:", df.columns)
    password_column = "Password"
    df["Valid"] = df[password_column].apply(check_password)

    print(df)

    output_file = "results/password_validation_detailed_results.csv"
    df.to_csv(output_file, index=False)
    
    print(f"The results have been saved to '{output_file}'.")

except FileNotFoundError:
    print("Error: The specified file was not found. Please check the file path.")
except KeyError:
    print("Error: The column containing the passwords was not found. Please check the exact column name.")
```

```python
#2-Password Validation with Yes/No Results
```

```python
def is_password_valid(password):
   
    if len(password) < 12:
        return "No"
   
    if not any(char.isupper() for char in password):
        return "No"
 
    if not any(char in "@()&" for char in password):
        return "No"
    # If all criteria are met, return "Yes"
    return "Yes"

file_path = "data/Candidate_Passwords.csv"
```

```python
try:
    df = pd.read_csv(file_path)

    print("Available columns:", df.columns)
    password_column = "Password"
    df["Result"] = df[password_column].apply(is_password_valid)

    # Create a new DataFrame with two columns: original password and result
    df_result = df[[password_column, "Result"]]

    # Save the results to an Excel file
    output_file = "results/password_validation_results.xlsx"
    df_result.to_excel(output_file, index=False)

    print(f"The results have been saved to '{output_file}'.")

except FileNotFoundError:
    print("Error: The specified file was not found. Please check the file path.")
except KeyError:
    print("Error: The column containing the passwords was not found. Please check the exact column name.")
```

```python
# CREDIT RISK AND LOAN ANALYSIS

# 1. Loan Amount Analysis by Purpose
```

```python
df = pd.read_excel("data/credit_loan.xlsx")
print(df.shape)
```

```python
file_path = "data/credit_loan.xlsx"

try:
    df = pd.read_excel(file_path)

    # Filter (loan_status = 1)
    df_approved = df[df['loan_status'] == 1]

    # stats for each loan_purpose
    stats = df_approved.groupby('loan_purpose')['loan_amnt'].describe(percentiles=[0.25, 0.75])

    # columns needed
    stats = stats[['min', '25%', 'mean', '75%', 'max']]

    # Export
    output_file = "results/loan_amount_statistics_by_purpose.xlsx"
    stats.to_excel(output_file)

    print(f"The statistics of approved loan amounts have been saved to '{output_file}'.")

except FileNotFoundError:
    print("Error: The specified file was not found. Please check the file path.")
except KeyError as e:
    print(f"Error: The following column was not found in the data: {e}")
```

```python
# 2. Interest Rate Analysis by Loan Purpose
```

```python
file_path = "data/credit_loan.xlsx"

try:
    
    df = pd.read_excel(file_path)

    # Filter (loan_status == 1)
    df_approved = df[df['loan_status'] == 1]

    # Calculate the range of interest rates for each loan purpose
    interest_rate_range = df_approved.groupby('loan_purpose')['loan_int_rate'].agg(['min', 'max'])

    interest_rate_range['range'] = interest_rate_range['max'] - interest_rate_range['min']

    # Export
    output_file = "results/interest_rate_analysis_by_loan_purpose.xlsx"

    interest_rate_range.to_excel(output_file)

    print(f"The ranges of approved interest rates have been saved to '{output_file}'.")

except FileNotFoundError:
    print("Error: The specified file was not found. Please check the file path.")

except KeyError as e:
    print(f"Error: The following column was not found in the data: {e}")
```

```python
# 3. Credit Score Analysis by Default History
```

```python
file_path = "data/credit_loan.xlsx"

try:
    df = pd.read_excel(file_path)

    # Group data based on the history of defaults ("previous_loan_defaults_on_file")
    # Separate data for people with and without a history of defaults
    group_defaults = df.groupby('previous_loan_defaults_on_file')['credit_score']

    #basic statistics for each group
    stats_defaults = group_defaults.describe()

    output_file = "results/credit_score_analysis_by_default_history.xlsx"
    stats_defaults.to_excel(output_file)

    print(f"The credit score statistics for people with and without a history of defaults have been saved to '{output_file}'.")

except FileNotFoundError:
    print("Error: The specified file was not found. Please check the file path.")

except KeyError as e:
    print(f"Error: The following column was not found in the data: {e}")

except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

```python
# 4. Loan Approval Analysis by Default History
```

```python
file_path = "data/credit_loan.xlsx"

try:
    
    df = pd.read_excel(file_path)

    # (loan_status == 1)
    loan_approval_counts = df.groupby('previous_loan_defaults_on_file')['loan_status'].value_counts().unstack(fill_value=0)

    #successfully obtained loans
    approved_with_defaults = loan_approval_counts.loc['Yes', 1] if 'Yes' in loan_approval_counts.index and 1 in loan_approval_counts.columns else 0

    # loans that were rejected
    rejected_defaults = loan_approval_counts.loc['Yes', 0] if 'Yes' in loan_approval_counts.index and 0 in loan_approval_counts.columns else 0

    # the count of loan approvals for individuals without default records
    approved_without_defaults = loan_approval_counts.loc['No', 1] if 'No' in loan_approval_counts.index and 1 in loan_approval_counts.columns else 0

    summary = pd.DataFrame({
        'Loan Type': ['Approved with defaults', 'Rejected defaults', 'Approved without defaults'],
        'Count': [approved_with_defaults, rejected_defaults, approved_without_defaults]
    })

    # Export
    output_file = "results/loan_approval_analysis_by_default_history.xlsx"
    summary.to_excel(output_file, index=False, engine='openpyxl')

    print(f"Loan approval data has been successfully saved to '{output_file}'.")

except FileNotFoundError:
    print("Error: The specified file was not found. Please check the file path.")

except KeyError as e:
    print(f"Error: A necessary column is missing in the data: {e}")

except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

```python
# 5. Average Approved Loan Amount for High Credit Scores
```

```python
file_path = "data/credit_loan.xlsx"

try:
    df = pd.read_excel(file_path)

    # Filter(loan_status == 1) and (credit_score >= 700)
    df_good_credit = df[(df['loan_status'] == 1) & (df['credit_score'] >= 700)]

    # Group by loan purpose and calculate the average approved loan amount
    mean_loan_amount = df_good_credit.groupby('loan_purpose')['loan_amnt'].mean()

    # Convert for easier export
    mean_loan_amount_df = mean_loan_amount.reset_index()
    mean_loan_amount_df.columns = ['Loan Purpose', 'Average Approved Loan Amount']

    #Export the result to an Excel file
    output_file = "results/average_approved_loan_amount_high_credit_scores.xlsx"
    mean_loan_amount_df.to_excel(output_file, index=False, engine='openpyxl')

    print(f"The data statistics for approved loan amounts for individuals with good credit have been saved to '{output_file}'.")

except FileNotFoundError:
    print("Error: The specified file was not found. Please check the file path.")
except KeyError as e:
    print(f"Error: The following column was not found in the data: {e}")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
